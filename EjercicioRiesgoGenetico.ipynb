```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
```

```python
df_riesgo = pd.read_csv('genetic_risk_dataset.csv')
df_riesgo.shape
```

```python
df_riesgo.dtypes
```

```python
print(df_riesgo.isna().sum())
```

```python
##Rellenando de forma mas rapida cada columna

columnas_numericas = df_riesgo.select_dtypes(include = np.number).columns
columnas_categoricas = df_riesgo.select_dtypes(exclude = np.number).columns

imp_num = SimpleImputer(strategy = "mean")
df_riesgo[columnas_numericas] = imp_num.fit_transform(df_riesgo[columnas_numericas])

imp_cat = SimpleImputer(strategy="most_frequent")
df_riesgo[columnas_categoricas] = imp_cat.fit_transform(df_riesgo[columnas_categoricas])
```

```python
##puedo sacar el IQ restando el 75% y el 25%
#columnas_ordinales = ['gene_expression_risk', 'enviromental_exposure']
df_codificado = pd.get_dummies(df_riesgo, columns = ['gene_expression_risk'], dtype = int)
df_codificado.head()
```

```python
df_coded = pd.get_dummies(df_codificado, columns = ['environmental_exposure'], dtype = int)
df_coded.head()
```

```python
Q1 = df_coded[['cholesterol', 'bmi', 'glucose', 'gene_marker_3', 'gene_marker_5']].quantile(0.25)
Q3 = df_coded[['cholesterol', 'bmi', 'glucose', 'gene_marker_3', 'gene_marker_5']].quantile(0.75)
IQR = Q3-Q1
```

```python
df_sin_atipicos = df_coded[~(df_coded[['cholesterol', 'bmi', 'glucose', 'gene_marker_3', 'gene_marker_5']] > (Q1 - 1.5 * IQR)) | (df_coded[['cholesterol', 'bmi', 'glucose', 'gene_marker_3', 'gene_marker_5']] > (Q3 +1.5 * IQR))]
```
